```python
from pytrends.request import TrendReq
import pandas as pd
```

```python
pytrends = TrendReq(
    hl='en-US',
    tz=360
)

keywords = [
    "Rolex Daytona",
    "Rolex Submariner",
    "Rolex GMT-Master II"
]

pytrends.build_payload(
    kw_list=keywords,
    timeframe='today 12-m'
)

df_trends = pytrends.interest_over_time()

df_trends.head()
```

```python
if "isPartial" in df_trends.columns:
    df_trends = df_trends.drop(columns=["isPartial"])

df_trends.to_csv(
    "../data/processed/google_trends_rolex.csv"
)

print("Saved to ../data/processed/google_trends_rolex.csv")
```
