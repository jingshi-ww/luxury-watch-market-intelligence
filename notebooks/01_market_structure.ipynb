```python
#segmentation
# Create a working dataframe with usable price and year
df_analysis = df[
    df['price_clean'].notna() &
    df['yop_clean'].notna()
].copy()

# Keep realistic watch production years
df_analysis = df_analysis[
    (df_analysis['yop_clean'] >= 1900) &
    (df_analysis['yop_clean'] <= 2024)
].copy()

# Create decade column
df_analysis['decade'] = (
    (df_analysis['yop_clean'] // 10) * 10
).astype(int).astype(str) + 's'

df_analysis[['yop', 'yop_clean', 'decade']].head(20)
```

```python
decade_volume = (
    df_analysis['decade']
    .value_counts()
    .sort_index()
)

decade_volume
```

```python
import plotly.express as px

decade_volume_df = decade_volume.reset_index()
decade_volume_df.columns = ['decade', 'listing_count']

fig = px.bar(
    decade_volume_df,
    x='decade',
    y='listing_count',
    title='Listing Volume by Production Decade',
    labels={
        'decade': 'Production Decade',
        'listing_count': 'Listing Volume'
    }
)

fig.update_layout(
    width=1000,
    height=550,
    plot_bgcolor='white',
    paper_bgcolor='white'
)

fig.show()
```

```python
decade_price = (
    df_analysis
    .groupby('decade')['price_clean']
    .median()
    .sort_index()
)

decade_price
```

```python
decade_price_df = decade_price.reset_index()
decade_price_df.columns = ['decade', 'median_price']

fig = px.bar(
    decade_price_df,
    x='decade',
    y='median_price',
    title='Observed Market Median Price Across Production Decades',
    labels={
        'decade': 'Production Decade',
        'median_price': 'Median Price (USD)'
    }
)

fig.update_layout(
    width=1000,
    height=550,
    plot_bgcolor='white',
    paper_bgcolor='white'
)

fig.show()
```

The observed price increase across decades may partly reflect shifts in brand composition and the growing dominance of high-end modern luxury brands.

While newer production decades appear associated with higher market median prices, this trend should not be interpreted solely as a vintage-versus-modern valuation effect.

The evolution of market composition also plays a significant role. Recent decades show a growing concentration of high-end luxury sports watch brands such as Rolex, Audemars Piguet, and Patek Philippe, which likely contributes to the increase in observed market median prices.

```python
import pandas as pd
import matplotlib.pyplot as plt

selected_brands = [
    'Rolex',
    'Omega',
    'Cartier',
    'Seiko',
    'Breitling',
    'Longines',
    'Audemars Piguet',
    'Patek Philippe'
]

df_brand = df_analysis[
    df_analysis['brand'].isin(selected_brands)
].copy()

brand_decade_count = (
    df_brand
    .groupby(['decade', 'brand'])
    .size()
    .unstack(fill_value=0)
)

# Keep decade order
brand_decade_count = brand_decade_count.sort_index()

median_price = (
    df_analysis
    .groupby('decade')['price_clean']
    .median()
    .reindex(brand_decade_count.index)
)

fig, ax1 = plt.subplots(figsize=(16, 8))

brand_decade_count.plot(
    kind='bar',
    stacked=True,
    ax=ax1,
    alpha=0.85
)

ax1.set_title(
    'Brand Composition and Market Median Price by Production Decade',
    fontsize=20,
    fontweight='bold',
    pad=20
)

ax1.set_xlabel('Production Decade', fontsize=14)
ax1.set_ylabel('Listing Volume', fontsize=14)
ax1.tick_params(axis='x', rotation=0)

ax2 = ax1.twinx()

ax2.plot(
    median_price.index,
    median_price.values,
    color='black',
    marker='o',
    linewidth=3,
    label='Median Price'
)

ax2.set_ylabel('Median Market Price (USD)', fontsize=14)

# Legends
ax1.legend(
    title='Brand',
    bbox_to_anchor=(1.08, 1),
    loc='upper left'
)

ax2.legend(
    bbox_to_anchor=(1.08, 0.55),
    loc='upper left'
)

plt.tight_layout()
plt.show()
```
